```python
salaries = [28000,40000,26000,48000,
            32000,36000,40000,42000,
            36000,28000,15000,]
```

```python
total_salary = sum(salaries)
average_salary = round(total_salary / len(salaries) , 2)
highest_salary = max(salaries)
lowest_salary = min(salaries)
no_of_employes = len(salaries)
```

```python
print(f"Number of employees:{no_of_employes} ")
print(f"salaries are :{salaries}")
print(f"Total salary: ₹{total_salary}")
print(f"Average salary: ₹{average_salary}")
print(f"Highest salary: ₹{highest_salary}")
print(f"Lowest salary: ₹{lowest_salary}")
```
